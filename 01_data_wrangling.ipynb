# 1. Data collection and wrangling
This notebook loads the course datasets, inspects their structure, and prepares a consistent dataframe. Run `python src/download_data.py` from the project root first.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
if not (ROOT / "data").exists() and (ROOT.parent / "data").exists():
    ROOT = ROOT.parent
DATA = ROOT / "data"
print("Project root:", ROOT)
print("Available CSVs:", [p.name for p in DATA.glob("*.csv")])
```

```python
def load_csv(filename):
    path = DATA / filename
    if not path.exists():
        raise FileNotFoundError(f"{path} not found. Run: python src/download_data.py")
    return pd.read_csv(path)

available = {p.name for p in DATA.glob("*.csv")}
if "dataset_part_1.csv" in available:
    raw = load_csv("dataset_part_1.csv")
elif "dataset_part_2.csv" in available:
    raw = load_csv("dataset_part_2.csv")
else:
    raise FileNotFoundError("Expected dataset_part_1.csv or dataset_part_2.csv in data/.")
display(raw.head())
print("Shape:", raw.shape)
display(raw.info())
```

```python
print("Missing values:")
display(raw.isna().sum().sort_values(ascending=False).head(15))
print("Duplicate rows:", raw.duplicated().sum())
```

```python
df = raw.drop_duplicates().copy()
# Standardize column names without destroying meaningful labels.
df.columns = [c.strip() for c in df.columns]
print("Cleaned shape:", df.shape)
display(df.describe(include="all").T.head(20))
```
