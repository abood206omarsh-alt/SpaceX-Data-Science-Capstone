# 2. Exploratory data analysis and visualization
Explore the course dataset and generate plots from the actual data loaded on your machine.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
ROOT = Path.cwd()
if not (ROOT / "data").exists() and (ROOT.parent / "data").exists(): ROOT = ROOT.parent
DATA = ROOT / "data"
file = next((DATA / n for n in ["dataset_part_2.csv", "dataset_part_1.csv", "spacex_launch_dash.csv"] if (DATA/n).exists()), None)
if file is None: raise FileNotFoundError("Run src/download_data.py first.")
df = pd.read_csv(file)
print("Using:", file.name, "| shape:", df.shape)
display(df.head())
```

```python
print("Columns:", df.columns.tolist())
display(df.isna().sum().sort_values(ascending=False).head(15))
```

```python
site_col = next((c for c in ["Launch Site", "LaunchSite", "launch_site"] if c in df.columns), None)
if site_col:
    plt.figure(figsize=(9, 4))
    df[site_col].value_counts().plot(kind="bar")
    plt.title("Number of launches by launch site")
    plt.xlabel("Launch site"); plt.ylabel("Launches"); plt.xticks(rotation=30, ha="right"); plt.tight_layout(); plt.show()
else:
    print("Launch-site column not found in this dataset version.")
```

```python
target_col = next((c for c in ["Class", "class", "LandingOutcome", "Outcome"] if c in df.columns), None)
if target_col:
    plt.figure(figsize=(6, 4))
    df[target_col].astype(str).value_counts().plot(kind="bar")
    plt.title("Distribution of landing outcomes")
    plt.xlabel("Outcome"); plt.ylabel("Rows"); plt.tight_layout(); plt.show()
    display(df[target_col].astype(str).value_counts(normalize=True).rename("proportion"))
else:
    print("Outcome column not found; inspect the available columns above.")
```

```python
payload_col = next((c for c in ["Payload Mass (kg)", "PayloadMass", "payload_mass_kg"] if c in df.columns), None)
if payload_col and site_col:
    plt.figure(figsize=(10, 5))
    sns.boxplot(data=df, x=site_col, y=payload_col)
    plt.title("Payload mass by launch site")
    plt.xticks(rotation=30, ha="right"); plt.tight_layout(); plt.show()
else:
    print("Payload or launch-site column unavailable for this plot.")
```
